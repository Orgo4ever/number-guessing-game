```python
import random

# Set the outer limits for the game
ABSOLUTE_MIN_NUMBER = 1
ABSOLUTE_MAX_NUMBER = 100

# Set the smallest allowed size of the guessing range
MIN_RANGE_SIZE = 20

# Set the maximum number of valid guesses
MAX_GUESSES = 10

# Get and validate the player's name
while True:
    player_name = input("Enter your name: ").strip()

    if player_name == "":
        print("Name cannot be empty. Please enter your name.")
    else:
        break

# Randomly choose the minimum number for this round
min_number = random.randint(
    ABSOLUTE_MIN_NUMBER,
    ABSOLUTE_MAX_NUMBER - MIN_RANGE_SIZE
)

# Randomly choose the maximum number for this round
max_number = random.randint(
    min_number + MIN_RANGE_SIZE,
    ABSOLUTE_MAX_NUMBER
)

# Randomly choose the secret number inside the randomized range
secret_number = random.randint(min_number, max_number)

# Create the game state
game_state = {
    "player_name": player_name,
    "absolute_min_number": ABSOLUTE_MIN_NUMBER,
    "absolute_max_number": ABSOLUTE_MAX_NUMBER,
    "min_number": min_number,
    "max_number": max_number,
    "secret_number": secret_number,
    "min_range_size": MIN_RANGE_SIZE,
    "max_guesses": MAX_GUESSES,
    "guesses_taken": 0,
    "guesses_remaining": MAX_GUESSES,
    "guess_history": [],
    "last_guess": None,
    "won": False,
    "game_over": False
}

print(f"\nWelcome, {game_state['player_name']}!")
print(
    f"I am thinking of a number between "
    f"{game_state['min_number']} and {game_state['max_number']}."
)
print(f"You have {game_state['max_guesses']} guesses to find it.\n")

# Main game loop
while not game_state["game_over"]:

    try:
        guess = int(input("Enter your guess: "))

        if guess < game_state["min_number"] or guess > game_state["max_number"]:
            print(
                f"Please enter a number between "
                f"{game_state['min_number']} and {game_state['max_number']}."
            )
            continue

    except ValueError:
        print("Invalid input. Please enter a whole number.")
        continue

    # Update the game state after a valid guess
    game_state["last_guess"] = guess
    game_state["guess_history"].append(guess)
    game_state["guesses_taken"] += 1
    game_state["guesses_remaining"] = (
        game_state["max_guesses"] - game_state["guesses_taken"]
    )

    # Check the player's guess
    if guess == game_state["secret_number"]:
        game_state["won"] = True
        game_state["game_over"] = True
        print(
            f"\nCorrect! You guessed the number in "
            f"{game_state['guesses_taken']} guesses."
        )

    elif guess > game_state["secret_number"]:
        print("Too high.")

    else:
        print("Too low.")

    # Check whether the player has used all guesses
    if game_state["guesses_taken"] == game_state["max_guesses"] and not game_state["won"]:
        game_state["game_over"] = True
        print(f"\nGame over! The number was {game_state['secret_number']}.")

    # Show guesses remaining if the game is still running
    if not game_state["game_over"]:
        print(f"Guesses remaining: {game_state['guesses_remaining']}\n")

# Final game summary
print("\nGame Summary")
print("------------")
print(f"Player: {game_state['player_name']}")
print(
    f"Range: {game_state['min_number']} to {game_state['max_number']}"
)
print(f"Secret number: {game_state['secret_number']}")
print(f"Guesses taken: {game_state['guesses_taken']}")
print(f"Guesses: {game_state['guess_history']}")

if game_state["won"]:
    print("Result: Win")
else:
    print("Result: Loss")
```
